```python
import pandas as pd

# The test split holds 800 images each of digits 3, 5 and 8 (2,400 ids)
TEST_SIZE = 3 * 800

# Output constant prediction of 3 (resulting in 33.3% accuracy)
df_sub = pd.DataFrame({"id": list(range(TEST_SIZE)), "prediction": [3] * TEST_SIZE})
df_sub.to_parquet("submission.parquet")
```
